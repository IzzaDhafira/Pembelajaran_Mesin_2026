**Kegunaan masing-masing library**

1. PyPrep: Pustaka Python untuk pra-pemrosesan data Electroencephalography (EEG). Library ini berfungsi untuk mengidentifikasi saluran yang buruk, mengurangi noise referensi, dan membersihkan artefak dari sinyal aktivitas otak sebelum data dimasukkan ke dalam model machine learning.
2. Scipy: Pustaka fundamental untuk komputasi ilmiah dan teknis. Library ini menyediakan algoritma yang sangat penting untuk melakukan penyaringan frekuensi (filtering), transformasi sinyal, integrasi, optimasi, dan fungsi aljabar linier.
3. wandb: Platform pelacakan eksperimen machine learning. Library ini secara otomatis memantau proses pelatihan model, mencatat metrik evaluasi secara berkala, menyimpan parameter model, dan memvisualisasikan hasil performa algoritma secara real-time di dalam satu dashboard.
4. pyECG: Pustaka khusus untuk ekstraksi dan analisis fitur sinyal Electrocardiogram (ECG).

Jawab:
- Dampak Lingkungan: Perkembangan AI menuntut spesifikasi perangkat keras (GPU, TPU, dan server klaster) yang sangat tinggi. Karena arsitektur model AI terus berkembang pesat, siklus usia pakai (lifespan) perangkat keras keras tersebut menjadi sangat singkat atau cepat usang. Hal ini menghasilkan lonjakan Limbah Elektronik (E-waste). Komponen server AI mengandung material langka dan logam berat beracun (seperti timbal, kadmium, dan merkuri) yang jika dibuang tanpa pengolahan khusus akan merusak ekosistem dan mencemari sumber air tanah.
- Cara mengatasi: Menurut pendapat saya, penyelesaian masalah ini bisa dilakukan melalui efisiensi perangkat lunak dan arsitektur data. Cara pertama, kita harus mengoptimalkan pipeline data dengan menyajikan dataset yang sudah bersih, tervalidasi, dan tereduksi dimensinya, agar proses data mining tidak akan membuang siklus komputasi secara sia-sia, sehingga memperpanjang umur perangkat keras. Cara kedua, memaksimalkan adopsi infrastruktur Cloud Computing agar sumber daya server dapat digunakan bersama. Cara ketiga, mendukung regulasi yang mewajibkan perusahaan penyedia GPU menerapkan ekonomi sirkular untuk mendaur ulang komponen bekas mereka.
- Referensi:
- Forti V., Baldé C.P., Kuehr R., & Bel G. (2020). The Global E-waste Monitor 2020: Quantities, flows, and the circular economy potential. United Nations University (UNU) / ITU.
- Ligozat, A. L., Lefevre, J., Bugeau, A., & Combaz, J. (2022). Unraveling the hidden environmental impacts of AI solutions: Life cycle assessment of AI solutions. Sustainability.
